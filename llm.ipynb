```python
import re
def read_whatsapp(file_path):
    edited_message="<This message was edited>"
    deleted_message="You deleted this message"
    null_message="null";
    created="created group";
    added_you="added you"

    
    filtered_lines=[]
    with open(file_path, 'r') as f:
        lines=f.readlines()
        for line in lines:
            if(edited_message not in line and deleted_message not in line and null_message not in line and created not in line and added_you not in line):
                line=re.sub(r'^\d{1,2}/\d{1,2}/\d{2}, \d{1,2}:\d{2}\s*[ap]m - [^:]+: ', '', line, flags=re.MULTILINE)
                filtered_lines.append(line)
    with open("clean_chat.txt", 'w') as f:
        for i in range(1,len(filtered_lines)):
            f.write(filtered_lines[i])


read_whatsapp(r"C:\Users\FASI OWAIZ AHMED\Desktop\llm\WhatsApp Chat with AB_naughty.txt")
```

```python
#tokenizing
from tokenizers import Tokenizer, models, trainers, pre_tokenizers

# Create a BPE tokenizer
tokenizer = Tokenizer(models.BPE())
tokenizer.pre_tokenizer = pre_tokenizers.Whitespace()

trainer = trainers.BpeTrainer(vocab_size=30000, special_tokens=["[PAD]", "[UNK]", "[CLS]", "[SEP]", "[MASK]"])

# Load your cleaned corpus
files = [r"C:\Users\FASI OWAIZ AHMED\Desktop\llm\clean_chat.txt"]
tokenizer.train(files, trainer)

# Save the tokenizer
tokenizer.save("my_custom_tokenizer.json")
```
